```python
def f1(x):
 return x**3-5*x**2-3*x+5
 
```

```python
def der(x):
    return 3*x**2-10*x-3
```

```python
def f(x):
    return x**2-6*x+4
```

```python
from scipy.optimize import newton
newton(f=f1, fprime = der, x0=0, max_iter=20)
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-30,30,300)
y=f1(x)

plt.axhline(0)
plt.plot(x,y)
plt.xlim(-2.5,6)
plt.ylim(-10,10)
plt.show
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-30,30,300)
y=f(x)

plt.axhline(0)
plt.plot(x,y)
plt.xlim(-2.5,6)
plt.ylim(-10,10)
plt.show
```
